```python
#Carga de datos
import pandas as pd
import numpy as np
def CargadataFrames(data):
    df= pd.read_csv(data)
    return df

ventas=CargadataFrames("ventas.csv")
marketing=CargadataFrames("marketing.csv")
clientes=CargadataFrames("clientes.csv")

#Forma que tiene el dataFrame
print("ventas:", ventas.shape)
print("clientes:", clientes.shape)
print("marketing:", marketing.shape)
```

```python
#Exploración de los datos
import pandas as pd

# Mostrar un resumen estadístico según tipo de dato específico
def resumen_por_tipo(df,nombre,tipo):
  variable = df.select_dtypes(include=tipo)
  if not variable.empty:
      print(f"Resumen del tipo {tipo} de {nombre}:")
      display(variable.describe().T)
  else:
        print(f"No hay columnas del tipo {tipo} en {nombre}")

#Funcion que explora los datos de un dataframe
def explorar_Datos(df, nombre):
    print(f"{'='*60}")
    print(f"EXPLORACIÓN INICIAL DE: {nombre.upper()}")
    print(f"{'='*60}")

    #Dimensiones y tipos
    print("Dimensiones:")
    print(f"Filas: {df.shape[0]} | Columnas: {df.shape[1]}")
    print("\nTipos de datos:")
    for tipo, cantidad in df.dtypes.items():
        print(f"{tipo}:        {cantidad}")

    #Calidad de datos
    print("\nCALIDAD DE DATOS")
    nulos=df.isna().sum()
    print("\nCantidad de Nulos\n")
    print(nulos)
    duplicados = df.duplicated().sum()
    print(" ")
    print(f"Cantidad de duplicados->{duplicados}")
    print(f"{'='*60}")

    #Vista rápida de los datos
    print(f"\nPrimeras filas de {nombre}:")
    display(df.head(5))

    # Estadísticas básicas (Numerico o Categórico)
    print(f"\n{'='*60}")
    print("RESUMEN NUMÉRICO:")
    print(f"{'='*60}")
    resumen_por_tipo(df,nombre,'number')
    print(f"\n{'='*60}")
    print("RESUMEN CATEGÓRICO:")
    print(f"{'='*60}")
    resumen_por_tipo(df,nombre,'object')
    print("\nFin de la exploración.")
```

```python
#Exploramos cada Dataset
explorar_Datos(ventas,"ventas")
print(f"\n{'---'*60}")
explorar_Datos(clientes,"clientes")
print(f"\n{'---'*60}")
explorar_Datos(marketing,"marketing")
```

# **Limpieza de datos**
La limpieza de datos es una de las etapas más cruciales dentro del proceso de análisis, ya que garantiza que la información utilizada sea precisa, coherente y confiable para la toma de decisiones. En la práctica, los datos suelen presentarse con errores, valores nulos, duplicados o inconsistencias que pueden comprometer la calidad del análisis y conducir a conclusiones incorrectas si no se corrigen adecuadamente.

Durante el proceso realizado anteriormente, se detectaron valores nulos, registros duplicados y formatos inconsistentes en precios y fechas, los cuales requirieron una serie de procedimientos de limpieza. Además, se normalizará los valores monetarios para convertirlos de texto a formato numérico y se parsearon las fechas en las tablas ventas y marketing utilizando métodos de conversión automáticos (pd.to_datetime) que aseguran una interpretación uniforme de los días y meses, evitando errores en el orden o formato.

```python
#Limpieza de datos
#Saber que datos hay duplicados
def EstadoDtos(df,nombre):
  estado_inicial = {
    "Filas y columnas": df.shape,
    "Nulos en precio": df["precio"].isna().sum(),
    "Nulos en cantidad": df["cantidad"].isna().sum(),
    "Duplicados": df.duplicated().sum()
    }
# Mostrar el resumen
  print(f"========== ESTADO INICIAL DE LOS DATOS EN {nombre.upper()} ==========")
  for clave, valor in estado_inicial.items():
    print(f"{clave}: {valor}")
  print("=================================================")

EstadoDtos(ventas,"ventas")
```

```python
#Proceso de normalización
# 1) Normalizar precio (De string a float)
if 'precio' in ventas.columns:
    ventas['precio'] = pd.to_numeric(
        ventas['precio'].replace(r'[^0-9.]', '', regex=True),
        errors='coerce'
    )
# 2) Parsear fechas en "ventas" y "marketing"
# En ventas
if 'fecha_venta' in ventas.columns:
    ventas['fecha_venta'] = ventas['fecha_venta'].apply(
        lambda x: pd.to_datetime(x, dayfirst=True, errors='coerce')
    )

# En marketing
for col in ['fecha_inicio', 'fecha_fin']:
    if col in marketing.columns:
        marketing[col] = marketing[col].apply(
            lambda x: pd.to_datetime(x, dayfirst=True, errors='coerce')
        )
# 3) Eliminar filas con nulos críticos en "ventas"
columnas_criticas = ['precio', 'cantidad']

mask_validas = ventas[columnas_criticas].notna().all(axis=1) # Crear máscara de filas válidas


filas_eliminadas = (~mask_validas).sum() # Contar filas que serán eliminadas


ventas = ventas[mask_validas].reset_index(drop=True) # Filtrar DataFrame y resetear índice

# 4) Eliminar duplicados en ventas
len_antes = len(ventas)
ventas = ventas.drop_duplicates().copy()
len_despues = len(ventas)
dup_eliminados = len_antes - len_despues
#print("Duplicados eliminados:", dup_eliminados)

# 5) Crear importe
ventas['importe'] = ventas['precio'] * ventas['cantidad']
```

```python
# ----------------------------------------------------
# VISUALIZACIÓN DEL ESTADO DE NORMALIZACIÓN
# ----------------------------------------------------
print("="*60)
print("RESUMEN DEL ESTADO DE NORMALIZACIÓN")
print("="*60)

# Normalización de precio
if 'precio' in ventas.columns:
    n_nulos_precio = ventas['precio'].isna().sum()
    print("\n1) Normalización de 'precio'")
    print(f" - Tipo de dato: {ventas['precio'].dtype}")
    print(f" - Valores nulos después de normalizar: {n_nulos_precio}")
    print(f" - Primeros 3 valores: {ventas['precio'].head(3).tolist()}")

#Eliminación de nulos críticos
print("\n3) Eliminación de filas con nulos críticos en ventas")
print(f" - Columnas críticas: {columnas_criticas}")
print(f" - Filas eliminadas: {filas_eliminadas}")
print(f" - Total de filas después de limpieza: {len(ventas)}")

#Eliminación de duplicados
print("\n4) Eliminación de duplicados en ventas")
print(f" - Duplicados eliminados: {dup_eliminados}")
print(f" - Total de filas después de eliminar duplicados: {len(ventas)}")

#Creación de columna 'importe'
print("\n5) Creación de columna 'importe'")
print(f" - Tipo de dato: {ventas['importe'].dtype}")
print(f" - Primeros 3 valores de 'importe': {ventas['importe'].head(3).tolist()}")

print("\nFIN DEL RESUMEN DE NORMALIZACIÓN")
print("="*60)
```

```python
#Etapa 1 - Actividad 2 "Realizar un script básico que calcule las ventas mensuales utilizando variables y operadores"

mes= { # Diccionario de nombres de meses
    1:"Enero", 2:"Febrero", 3:"Marzo", 4:"Abril", 5:"Mayo", 6:"Junio",
    7:"Julio", 8:"Agosto", 9:"Septiembre", 10:"Octubre", 11:"Noviembre", 12:"Diciembre"}

columnas_requeridas = {"fecha_venta", "importe"} #Validaciones iniciales
if not columnas_requeridas.issubset(ventas.columns):
    print("-"*40)
    print("ERROR: Faltan columnas requeridas:", columnas_requeridas - set(ventas.columns))
    print("-"*40)
else:# Limpieza y validación de datos
    filas_totales = len(ventas)
    ventas_validas = ventas.dropna(subset=["fecha_venta", "importe"]).copy()
    filas_omitidas = filas_totales - len(ventas_validas)

    # Convertir fecha si aún no está en formato datetime
    if not np.issubdtype(ventas_validas["fecha_venta"].dtype, np.datetime64):
        ventas_validas["fecha_venta"] = pd.to_datetime(ventas_validas["fecha_venta"], errors='coerce')

    # Eliminar importes no válidos
    ventas_validas = ventas_validas[ventas_validas["importe"] > 0]


    # Agrupar por año y mes
    ventas_validas["Año"] = ventas_validas["fecha_venta"].dt.year
    ventas_validas["Mes"] = ventas_validas["fecha_venta"].dt.month
    ventas_validas["Mes_Nombre"] = ventas_validas["Mes"].map(mes)

    # Calcular total mensual
    resumen = (
        ventas_validas
        .groupby(["Año", "Mes", "Mes_Nombre"], as_index=False)["importe"]
        .sum()
        .sort_values(["Año", "Mes"]) )


    # Mostrar resultados
    print("\n" + "-"*40)
    print("Ventas mensuales")
    print("-"*40)
    for _, fila in resumen.iterrows():
        print(f"{fila['Año']}-{fila['Mes']:02d} ({fila['Mes_Nombre']}): $ {fila['importe']:.2f}")
    print("\n" + "-"*40)
    print("Resumen de control")
    print(f"Filas procesadas: {len(ventas_validas)}")
    print(f"Filas omitidas : {filas_omitidas}")
    print("-"*40)

    display(resumen)
```

```python
# Etapa 1 - Actividad 3 Estructuras de Datos: Desarrollar un programa que almacene los datos de ventas (producto, precio, cantidad).

# Copia del DataFrame con solo las filas válidas
ventas_validas = ventas.dropna(subset=["producto", "precio", "cantidad"]).copy()

# Inicializar la lista donde guardaremos los registros
registros = []
for _, fila in ventas_validas.iterrows():
    registro = {
        "producto": fila["producto"],
        "precio": float(fila["precio"]),
        "cantidad": float(fila["cantidad"])
    }
    registros.append(registro)

#Resultado
print("Vista previa de registros normalizados:")
display(registros)
```

# ¿Que conviene más utilizar diccionarios o listas?

Usamos una lista de diccionarios porque cada fila del DataFrame representa un registro independiente y por cada registro necesita almacenar campos con nombre como “producto”, “precio” y “cantidad”.

Pero, un diccionario puede resolver este problema, si necesita representar una fila completa, ya que se puede acceder a los valores mediante claves, pero como existen muchas filas, necesitamos una estructura que agrupe todos esos diccionarios. Por eso, la lista reúne todos los registros y el diccionario organiza los datos de cada uno.


```python
# Etapa 2 - Actividad 2 Transformación de Datos: Aplicar filtros y transformaciones para
# crear una tabla de ventas que muestre solo los productos con alto rendimiento.

TOP_N = 10  # cantidad de productos con mayores ingresos

# Validar columnas necesarias
requeridas = {"producto", "importe"}
faltan = requeridas - set(ventas.columns)
if faltan:
    raise ValueError(f"Faltan columnas requeridas en 'ventas': {faltan}")

# Agrupar por producto y calcular ingresos totales
ingresos_por_producto = ventas.groupby("producto", as_index=False)["importe"].sum()
ingresos_por_producto.rename(columns={"importe": "ingresos_totales"}, inplace=True)

# Obtener directamente el Top N usando nlargest
top_por_ingresos = ingresos_por_producto.nlargest(TOP_N, "ingresos_totales")

# Mostrar resultados
print("Productos que más dinero generaron:")
display(top_por_ingresos)

```

# Etapa 2 - Actividad 3 "Agregación: Resumir las ventas por categoría de producto y analizar los ingresos generados"
En esta etapa voy a analizar la evolución mensual por categoría, ya que muestra cómo evolucionan los ingresos mensuales por categoría, permitiendo identificar la estacionalidad al detectar meses fuertes o débiles, reconocer picos o caídas puntuales en las ventas, y servir como base para generar gráficos y análisis rápidos de la evolución de cada categoría.

```python
# ----------------------------------------------------
# Top productos por ingresos
# ----------------------------------------------------
# Etapa 2 - Actividad 3: Agregación por Categoría
# Agrupar por categoria y sumar importes
ventas_por_categoria = ventas.groupby("categoria", as_index=False)["importe"].sum()

# Ordenar de mayor a menor venta
ventas_por_categoria = ventas_por_categoria.sort_values("importe", ascending=False)

print("Ingresos generados por Categoría:")
display(ventas_por_categoria)
```

# Etapa 2 - Actividad 4: Integración de Datos

```python
"""
Unimos con "Merge" el DataFrame de ventas con el de marketing usando 'producto' como clave,
usando how='left'para mantener todas las ventas,aunque no tengan campaña de marketing asociada.

Despues, rellenamos los valores nulos en 'costo' (productos sin marketing) con 0 para poder operar matemáticamente.
Y al finalizar, mostramos las primeras filas para confirmar que se agregaron las columnas de marketing y validamos si hay productos que se vendieron
y tuvieron costo de marketing
"""
dataset_final = pd.merge(ventas, marketing, on="producto", how="left")

dataset_final["costo"] = dataset_final["costo"].fillna(0)
dataset_final["canal"] = dataset_final["canal"].fillna("Sin Campaña")

# Mostrar las dimensiones del nuevo dataset consolidado
print(f"Dimensiones del Dataset Integrado: {dataset_final.shape}")

# Ver las primeras filas para confirmar que se agregaron las columnas de marketing (canal, costo, etc.)
print("Vista previa del Dataset Integrado:")
display(dataset_final.head())

# Validar: Verificar si hay productos que se vendieron y tuvieron costo de marketing
print("\nEjemplo de ventas con información de marketing asociada:")
display(dataset_final[dataset_final["costo"] > 0].head(3))
```

```python
# ==========================================
# CREACIÓN DE DATAFRAME MAESTRO POR PRODUCTO ----BASE DE TODO----
# ==========================================
import numpy as np
# 1. Agrupar Ventas por Producto
ventas_prod = ventas.groupby("producto").agg({
    "importe": "sum",
    "cantidad": "sum",
    "id_venta": "count" # Cuenta transacciones
}).reset_index()
ventas_prod.rename(columns={
    "importe": "ventas_total",
    "cantidad": "unidades_vendidas",
    "id_venta": "transacciones"
}, inplace=True)

# 2. Agrupar Marketing por Producto
mkt_prod = marketing.groupby("producto").agg({
    "costo": "sum",
    "id_campanha": "count"
}).reset_index()
mkt_prod.rename(columns={
    "costo": "inversion_total",
    "id_campanha": "campañas"
}, inplace=True)

# 3. Fusionar (Merge) para crear prod_global
prod_global = pd.merge(ventas_prod, mkt_prod, on="producto", how="left")

# Rellenar nulos con 0 (para productos que no tuvieron marketing)
prod_global["inversion_total"] = prod_global["inversion_total"].fillna(0)
prod_global["campañas"] = prod_global["campañas"].fillna(0)

# 4. Cálculo de KPIs "Indicadores Clave"
# Ticket Medio: Cuánto gastan en promedio por compra de ese producto
prod_global["ticket_medio"] = prod_global["ventas_total"] / prod_global["transacciones"]

# ROAS (Retorno de Inversión): Ventas / Inversión
# Usamos np.where para evitar división por cero
prod_global["ROAS"] = np.where(
    prod_global["inversion_total"] > 0,
    prod_global["ventas_total"] / prod_global["inversion_total"],
    0
)
# Agregamos la Categoría (recuperándola del df original)
cat_map = ventas.drop_duplicates("producto").set_index("producto")["categoria"]
prod_global["categoria"] = prod_global["producto"].map(cat_map)

print("Dimensiones de prod_global:", prod_global.shape)
display(prod_global.head())

# Guardar el dataset
prod_global.to_csv("resumen_productos.csv", index=False)
#print("Archivo 'resumen_productos.csv' guardado exitosamente.")
```

## 1. Síntesis de Hallazgos
Tras procesar y analizar los datos transaccionales integrados con las campañas de marketing, hemos detectado los siguientes patrones clave:

* **Productos Estrella:** El análisis de ingresos totales (Top 5) revela que los productos que lideran la facturación son **Lamparas de mesas** y **Auriculares**. Estos artículos representan la base financiera de este negocio.

* **Estacionalidad:** Al observar la evolución mensual, detectamos que las ventas alcanzan su pico máximo en el mes de **Mayo**, mientras que en **Junio** se observa una caída, lo que sugiere una fuerte estacionalidad en la demanda.

## 2. Eficiencia de Marketing (ROAS)
Al cruzar los datos de inversión publicitaria con las ventas efectivas, descubrimos que:

* **Relación Inversión-Venta:** El gráfico de dispersión indica que **"No existe una correlación clara, indicando gasto ineficiente"**; ya que el coeficiente de correlación entre el costo de marketing y el importe de ventas es cercano a cero (-0.004). Esto sugiere que el aumento en la inversión publicitaria no se traduce linealmente en un aumento de las ventas para todos los productos.

* **El Indicador Clave:** El cálculo del **ROAS (Retorno de Inversión Publicitaria)** resultó ser el indicador más revelador.
    * Productos con ROAS alto (muy eficientes): **Lámpara de mesa** con un ROAS de 5164.87.
    * Productos con ROAS bajo (inversión sin retorno): **Candelabro** con un ROAS de 759.65.

## 3. Recomendaciones para la Gerencia
Basado en los datos, sugerimos las siguientes acciones estratégicas:

1.  **Optimización de Presupuesto:** Reducir la inversión en los productos con bajo ROAS y reasignar ese presupuesto a los productos estrella para maximizar el retorno.
2.  **Gestión de Stock:** Asegurar disponibilidad de inventario para el mes de **Mayo** para evitar quiebres de stock durante el pico de demanda.
3.  **Estrategia de Precios:** Para los productos con muchas unidades vendidas pero bajo ticket medio, evaluar estrategias de *bundling* (venta en combos) para elevar el valor del carrito promedio.
